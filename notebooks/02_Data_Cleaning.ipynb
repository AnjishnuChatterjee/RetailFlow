```python
import pandas as pd
```

```python
df1 = pd.read_csv("../data/processed/retail_2009_2010.csv")
df2 = pd.read_csv("../data/processed/retail_2010_2011.csv")

df = pd.concat([df1, df2], ignore_index=True)

print(df.shape)
```

```python
# Remove missing Customer ID
df = df.dropna(subset=["Customer ID"])

# Remove cancelled invoices
df = df[~df["Invoice"].astype(str).str.startswith("C")]

# Remove invalid quantities
df = df[df["Quantity"] > 0]

# Remove invalid prices
df = df[df["Price"] > 0]

# Convert date
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
```

```python
# Revenue
df["Revenue"] = df["Quantity"] * df["Price"]

# Month
df["Month"] = df["InvoiceDate"].dt.month_name()

# Day of Week
df["DayOfWeek"] = df["InvoiceDate"].dt.day_name()

# Hour
df["Hour"] = df["InvoiceDate"].dt.hour

# Weekend
df["IsWeekend"] = (df["InvoiceDate"].dt.weekday >= 5).astype(int)

df.head()
```

```python
df.to_csv("../data/processed/cleaned_retail.csv", index=False)

print("✅ Cleaned dataset saved!")
```
