```python
import pandas as pd
```

```python
df1 = pd.read_excel(
    "../data/raw/online_retail_II.xlsx",
    sheet_name="Year 2009-2010",
    engine="calamine"
)

print(df1.shape)
df1.head()
```

```python
df1.to_csv(
    "../data/processed/retail_2009_2010.csv",
    index=False
)

print("✅ First CSV saved")
```

```python
df2 = pd.read_excel(
    "../data/raw/online_retail_II.xlsx",
    sheet_name="Year 2010-2011",
    engine="calamine"
)

print(df2.shape)
df2.head()
```

```python
df2.to_csv(
    "../data/processed/retail_2010_2011.csv",
    index=False
)

print("✅ Second CSV saved")
```

```python
df1 = pd.read_csv("../data/processed/retail_2009_2010.csv")
df2 = pd.read_csv("../data/processed/retail_2010_2011.csv")

df = pd.concat([df1, df2], ignore_index=True)

print(df.shape)
```

```python
df.head()

df.info()

df.describe()
```

```python
df.to_csv(
    "../data/processed/merged_retail.csv",
    index=False
)

print("✅ Merged dataset saved successfully!")
```
